## 1. Librerias

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency

sns.set_theme(style="whitegrid")
print("OK")
```

## 2. Funcion auxiliar

```python
def chi2_detalle(tabla, nombres_filas=None, nombres_cols=None):
    """tabla: array 2D de frecuencias observadas"""
    tabla = np.asarray(tabla, dtype=float)
    chi2, p, gl, esperados = chi2_contingency(tabla)
    n = tabla.sum()
    r, c = tabla.shape
    # Phi (2x2) o V de Cramer
    k = min(r - 1, c - 1)
    v = np.sqrt(chi2 / (n * k)) if k > 0 else np.nan
    phi = np.sqrt(chi2 / n) if (r == 2 and c == 2) else None

    print("=" * 60)
    print("Frecuencias observadas:")
    print(tabla)
    print("\nFrecuencias esperadas:")
    print(np.round(esperados, 2))
    print(f"\nchi2 = {chi2:.4f}")
    print(f"gl   = {gl}")
    print(f"p    = {p:.6e}")
    print(f"n    = {int(n)}")
    if phi is not None:
        print(f"Phi  = {phi:.4f}")
    print(f"V de Cramer = {v:.4f}")
    if p < 0.05:
        print("Decision (alpha=0.05): se RECHAZA H0 (hay evidencia de asociacion)")
    else:
        print("Decision (alpha=0.05): NO se rechaza H0")
    print("=" * 60)
    return chi2, p, gl, esperados, v
```

## 3. Ejemplo del apunte (genero x producto)

```python
tabla_ejemplo = np.array([[40, 20], [30, 50]])
chi2_detalle(tabla_ejemplo)
```

## 4. Verificacion Ejercicio 1 (canal x exito)

```python
tabla_ej1 = np.array([[25, 15], [20, 40]])
chi2_detalle(tabla_ej1)

# Barras agrupadas
df1 = pd.DataFrame(tabla_ej1, index=["Email", "Redes"], columns=["Exito", "No exito"])
df1.plot(kind="bar", figsize=(8, 4))
plt.title("Canal vs resultado de campana")
plt.ylabel("Frecuencia")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 5. Verificacion Ejercicio 5 (turno x defecto)

```python
tabla_ej5 = np.array([[10, 90], [25, 75]])
chi2_detalle(tabla_ej5)
```

## 6. Dataset tips: fumador vs momento del dia

```python
tips = sns.load_dataset("tips")
tab = pd.crosstab(tips["smoker"], tips["time"])
display(tab)
chi2_detalle(tab.values)

tab.plot(kind="bar", stacked=True, figsize=(7, 4))
plt.title("Fumador vs tiempo (comida)")
plt.ylabel("Conteo")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 7. Tabla mas grande: dia x tiempo

```python
tab2 = pd.crosstab(tips["day"], tips["time"])
display(tab2)
chi2_detalle(tab2.values)
```
